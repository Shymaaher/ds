## 4. Категоризация данных
    
Проведите категоризацию данных:
- Разделите все игры по оценкам пользователей и выделите такие категории: высокая оценка (от 8 до 10 включительно), средняя оценка (от 3 до 8, не включая правую границу интервала) и низкая оценка (от 0 до 3, не включая правую границу интервала).

```python
def categorize_user(score):
    if pd.isna(score):
        return None
    elif score >= 8 and score <= 10:
        return 'высокая'
    elif score >= 3 and score < 8:
        return 'средняя'
    elif score >= 0 and score < 3:
        return 'низкая'

df_actual['user_category'] = df_actual['user_score_numeric'].apply(categorize_user)
display(df_actual['user_category'].value_counts(dropna=False))
```

- Разделите все игры по оценкам критиков и выделите такие категории: высокая оценка (от 80 до 100 включительно), средняя оценка (от 30 до 80, не включая правую границу интервала) и низкая оценка (от 0 до 30, не включая правую границу интервала).

```python
def categorize_critic(score):
    if 80 <= score <= 100:
        return 'высокая'
    elif 30 <= score < 80:
        return 'средняя'
    elif 0 <= score < 30:
        return 'низкая'
    else:
        return 'без оценки'


df_actual['critic_category'] = df_actual['critic_score'].apply(categorize_critic)
display(df_actual['critic_category'].value_counts(dropna=False))
```

- После категоризации данных проверьте результат: сгруппируйте данные по выделенным категориям и посчитайте количество игр в каждой категории.

```python
display(df_actual.groupby('user_category')['user_score'].count())
```

```python
display(df_actual.groupby('critic_category')['user_score'].count())
```
